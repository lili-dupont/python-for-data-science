Definitions : (à clarifier + rajouter des trucs)

- Milieu fermé / ouvert :  une personne relève du milieu fermé lorsqu'elle fait l'objet d'une mesure avec écrou. A l'inverse, une personne qui est prise en charge par un service de suivi (CPAL puis SPIP) uniquement pour une ou plusieurs mesures sans écrou est alors considérée comme relevant du milieu ouvert. 
-  Détenus / non détenus : les détenus sont les personnes écrouées faisant l'objet d'une mesure privative de liberté à l'intérieur d'un établissement pénitentiaire. Les non détenus sont à l'extérieur d'un établissement pénitentiaire.
- L'écrou est une procédure administrative d'identification d'une personne détenue, matérialisée par l'attribution d'un numéro d'écrou au sein de l'établissement pénitentiaire, distinct du titre juridique justifiant l'incarcération. Parmi elleux ya détenus simples, détenus placés à l'extérieur = on leur file l'hébergement, et condamnés sous bracelet électronique non mobile = on leur file pas l'hébergement

#### Graphique 1 : Part des femmes parmi les écroués et les détenus

```python
# transformer les valeurs qui sont du texte en nombres 

for col in ["Part de femmes_ecroues", "Part de femmes_detenus"]:
    jointure[col] = (
        jointure[col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.replace("%", "", regex=False)
        .str.strip()
    )
    jointure[col] = pd.to_numeric(jointure[col], errors="coerce")

print(jointure[["Part de femmes_ecroues", "Part de femmes_detenus"]].dtypes)


# créer le graphique

import matplotlib.pyplot as plt

jointure = jointure.sort_values("Au 1er janvier")

#supprimer les lignes avec valeurs manquantes
ecroues = jointure.dropna(subset=["Part de femmes_ecroues"])
detenus = jointure.dropna(subset=["Part de femmes_detenus"])

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(ecroues["Au 1er janvier"], ecroues["Part de femmes_ecroues"], label="Femmes parmi les écroués")
ax.plot(detenus["Au 1er janvier"], detenus["Part de femmes_detenus"], label="Femmes parmi les détenus")

ax.set_title("Part des femmes parmi les écroués et les détenus")
ax.set_xlabel("Année (1er janvier)")
ax.set_ylabel("Part des femmes (%)")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

Clef de lecture : En 1990, 4,4 % des écroués étaient des femmes.

Interprétation :
1. La part de femmes parmi les écroués (et les détenus) est toujours très faible, avec un pic à 4,4% autour de 1990.
2. Depuis 2000, elle est plus ou moins stable autour de 3,6% (avec une hausse récente qui peut être une autre oscillation ou une nouvelle tendance).
3. La part de femmes parmi les écroués est presque parfaitement corrélée à celle des femmes parmi les détenus, mais plus élevée : les peines hors des établissements pénitentiaires semblent plus souvent privilégiées pour les femmes.

```python
import re

def commune(nom):
    nom = re.sub(r"\(.*?\)", "", str(nom)).strip()
    mots = nom.split()
    if mots and mots[0].isupper() and len(mots[0]) <= 4:
        mots = mots[1:]
    return " ".join(mots)

df_carte["commune"] = df_carte["Etablissement"].apply(commune)
print(df_carte[["Etablissement", "commune"]].to_string())
```

```python
import time
from pathlib import Path

import pandas as pd
import requests

URL = "https://data.geopf.fr/geocodage/search"
cache = Path("data/coordonnees_etablissements.csv")

corrections = {
    "BORDEAUX GRADIGNAN": "Gradignan",
    "POITIERS VIVONNE": "Vivonne",
    "LILLE LOOS SEQUEDIN": "Loos",
    "JOUX LA VILLE": "JOUX-LA-VILLE",
    "ORLEANS SARAN": "SARAN",
    "ST ETIENNE LA TALAUDIERE": "LA TALAUDIÈRE",
    "RHONE": "MEYZIEU",
    "LYON CORBAS": "CORBAS",
    "FLEURY MEROGIS": "Fleury-Mérogis",
    "SUD FRANCILIEN": "Réau",
    "CORBEIL ESSONNES": "CORBEIL-ESSONNES",
    "CAEN IFS": "CAEN",
    "MULHOUSE LUTTERBACH": "LUTTERBACH",
    "CHALONS EN CHAMPAGNE": "Châlons-en-Champagne",
    "NANCY MAXEVILLE": "NANCY",
    "TOULOUSE SEYSSES": "SEYSSES",
    "VILLENEUVE LES MAGUELONE": "Villeneuve-lès-Maguelone"
}

# 1. on repart toujours du tableau complet
df12 = dico_dfcsv["df12"]
masque_outremer = df12["Direction Interrégionale"].str.contains("outre", case=False, na=False)
df_carte = df12[~masque_outremer].copy().reset_index(drop=True)

# 2. communes, puis corrections
df_carte["commune"] = df_carte["Etablissement"].apply(commune)
df_carte["commune"] = df_carte["commune"].replace(corrections)

absents = [k for k in corrections if k not in df12["Etablissement"].apply(commune).values]
print("Clés de corrections qui ne correspondent à rien :", absents)

# 3. supprimer l'ancien cache (à retirer une fois les coordonnées validées)
cache.unlink(missing_ok=True)


def geocoder(commune):
    r = requests.get(URL, params={"q": commune, "type": "municipality", "limit": 1}, timeout=10)
    r.raise_for_status()
    feats = r.json().get("features", [])
    if not feats:
        return None, None, None, None
    lon, lat = feats[0]["geometry"]["coordinates"]
    p = feats[0]["properties"]
    return lat, lon, p.get("label"), p.get("score")


if cache.exists():
    coords = pd.read_csv(cache)
else:
    lignes = []
    for c in df_carte["commune"].dropna().unique():
        lat, lon, label, score = geocoder(c)
        lignes.append({"commune": c, "lat": lat, "lon": lon, "label": label, "score": score})
        time.sleep(0.1)
    coords = pd.DataFrame(lignes)
    cache.parent.mkdir(parents=True, exist_ok=True)
    coords.to_csv(cache, index=False)

print(coords.sort_values("score").head(10))

# 4. fusion sans doublon de colonnes
df_carte = df_carte.drop(columns=["lat", "lon", "label", "score"], errors="ignore")
df_carte = df_carte.merge(coords, on="commune", how="left")
print(df_carte["lat"].isna().sum(), "établissements sans coordonnées")
```

```python
import folium

for col in ["Places femmes", "Femmes détenues sur places femmes", "Taux d'occupation des places femmes"]:
    df_carte[col] = pd.to_numeric(
        df_carte[col].astype(str).str.replace(",", ".", regex=False).str.replace("%", "", regex=False).str.strip(),
        errors="coerce",
    )

carte = folium.Map(location=[46.6, 2.4], zoom_start=6)

for _, r in df_carte.dropna(subset=["lat", "lon", "Places femmes"]).iterrows():
    taux = r["Taux d'occupation des places femmes"]
    folium.CircleMarker(
        location=[r["lat"], r["lon"]],
        radius=max(r["Places femmes"] ** 0.5, 3),
        color="red" if taux > 100 else "steelblue",
        fill=True,
        fill_opacity=0.6,
        popup=f"{r['Etablissement']} : {taux:.0f} % ({r['Places femmes']:.0f} places)",
    ).add_to(carte)

carte
```
