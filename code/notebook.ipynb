## 1.1 Extraction des tableaux qui nous intéressent et conversion en .csv

```python
#Chargement des données
#ne pas relancer car les données sont déjà enregistrées dans le projet data/tableaux

import pandas as pd
chemin_donnees = "/home/onyxia/python-for-data-science/data/series_statistiques_ppsmj.xlsx"

chemin14 = "/home/onyxia/python-for-data-science/data/tableaux/strctr_parsexe_ecroues"
chemin15 = "/home/onyxia/python-for-data-science/data/tableaux/strctr_parsexe_detenus"
chemin52 = "/home/onyxia/python-for-data-science/data/tableaux/amengmts_peines_parsexe"
chemin60 = "/home/onyxia/python-for-data-science/data/tableaux/nb_prsn_suivies_milieuouvert_parsexe"
chemin67 = "/home/onyxia/python-for-data-science/data/tableaux/tx_prsn_ecrouees_parsexe_parage"
chemin68 = "/home/onyxia/python-for-data-science/data/tableaux/tx_prsn_detenues_parsexe_parage"

xls = pd.ExcelFile(chemin_donnees, engine="openpyxl")
#print(xls.sheet_names)          # vérifiez le nom exact de la feuille
df14 = xls.parse("Tableau 14")
df15 = xls.parse("Tableau 15")
df52 = xls.parse("Tableau 52")
df60 = xls.parse("Tableau 60")
dfsixseven = xls.parse("Tableau 67")
df68 = xls.parse("Tableau 68")

#df14.head

dico_dfcsv = {}
def to_csv(nom, df, chemin):
    df.to_csv(chemin, index=False, encoding="utf-8")
    dico_dfcsv[nom] = pd.read_csv(chemin)

to_csv("df14", df14, chemin14)
to_csv("df15", df15, chemin15)
to_csv("df52", df52, chemin52)
to_csv("df60", df60, chemin60)
to_csv("df67", dfsixseven, chemin67)
to_csv("df68", df68, chemin68)

#len(dico_dfcsv)
```

## 1.2. Visualisation des tableaux

Definitions : (à clarifier + rajouter des trucs)

- Milieu fermé / ouvert :  une personne relève du milieu fermé lorsqu'elle fait l'objet d'une mesure avec écrou. A l'inverse, une personne qui est prise en charge par un service de suivi (CPAL puis SPIP) uniquement pour une ou plusieurs mesures sans écrou est alors considérée comme relevant du milieu ouvert. 
-  Détenus / non détenus : les détenus sont les personnes écrouées faisant l'objet d'une mesure privative de liberté à l'intérieur d'un établissement pénitentiaire. Les non détenus sont à l'extérieur d'un établissement pénitentiaire.
- L'écrou est une procédure administrative d'identification d'une personne détenue, matérialisée par l'attribution d'un numéro d'écrou au sein de l'établissement pénitentiaire, distinct du titre juridique justifiant l'incarcération. Parmi elleux ya détenus simples, détenus placés à l'extérieur = on leur file l'hébergement, et condamnés sous bracelet électronique non mobile = on leur file pas l'hébergement
