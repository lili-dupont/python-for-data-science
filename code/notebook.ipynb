Definitions : (à clarifier + rajouter des trucs)

- Milieu fermé / ouvert :  une personne relève du milieu fermé lorsqu'elle fait l'objet d'une mesure avec écrou. A l'inverse, une personne qui est prise en charge par un service de suivi (CPAL puis SPIP) uniquement pour une ou plusieurs mesures sans écrou est alors considérée comme relevant du milieu ouvert. 
-  Détenus / non détenus : les détenus sont les personnes écrouées faisant l'objet d'une mesure privative de liberté à l'intérieur d'un établissement pénitentiaire. Les non détenus sont à l'extérieur d'un établissement pénitentiaire.
- L'écrou est une procédure administrative d'identification d'une personne détenue, matérialisée par l'attribution d'un numéro d'écrou au sein de l'établissement pénitentiaire, distinct du titre juridique justifiant l'incarcération. Parmi elleux ya détenus simples, détenus placés à l'extérieur = on leur file l'hébergement, et condamnés sous bracelet électronique non mobile = on leur file pas l'hébergement

## 1.3 Jointures et traitement des données

#### tableaux 1 et 2

```python
# Jointure des tableaux 1 et 2

print(dico_dfcsv["df1"].columns.tolist(), dico_dfcsv["df1"].shape)
print(dico_dfcsv["df2"].columns.tolist(), dico_dfcsv["df2"].shape)

'''
le tableau 2 ne commence qu'en 2005 donc il y a moins de lignes. on choisit de faire une jointure "outer" en laissant des données 
manquantes pour ensuite utiliser cette jointure pour construire un graphique dont une des courbes commence plus tard (Graphique 1)
'''

jointure = pd.merge(
    dico_dfcsv["df1"],
    dico_dfcsv["df2"],
    on="Au 1er janvier",
    how="outer",
    suffixes=("_ecroues", "_detenus"),
    validate="one_to_one",
)
print(jointure.shape)
jointure
```

#### tableaux 7 8 12

```python
# filtre du tableau 12
df12 = dico_dfcsv["df12"]
df12_filtre = df12[df12["Etablissement"].isin(["Total Direction Interrégionale","France entière"])]
df12_filtre.loc[65,"Direction Interrégionale"] = "Total France entière"

# filtre du tableau 7
df7 = dico_dfcsv["df7"]
df7_filtre = df7[df7["Direction Interrégionale"] != "Total Métropole"]

# filtre tableau 8
df8 = dico_dfcsv["df8"]
df8_filtre = df8[df8["Direction Interrégionale"] != "Total Métropole"]

# Vérification shape
#print(df12_filtre.columns.tolist(), df12_filtre.shape)
#print(df7_filtre.columns.tolist(), df7_filtre.shape)
#print(df8_filtre.columns.tolist(), df8_filtre.shape)
```

```python
# jointure tableaux 7, 8 et 12

# tableaux 8 et 7
jointure2 = pd.merge(
    df8_filtre,
    df7_filtre.drop(columns= ["Capacité norme circulaire", "Ecroués détenus"]), 
        # variable inutile (ratio calculé avec capacité opérationnelle) et variable redondante
    on="Direction Interrégionale",
    how="outer",
    validate="one_to_one",
)

# on rajoute tableau 12
jointure2 = pd.merge(
    jointure2,
    df12_filtre.drop(columns=["Etablissement", "Femmes détenues sur places femmes"]), # variable inutile et variable redondante
    on="Direction Interrégionale",
    how="outer",
    validate="one_to_one",
)
```

```python
# Conversion des variables en valeurs numériques
jointure2["Taux d'occupation des places femmes"] = pd.to_numeric(
    jointure2["Taux d'occupation des places femmes"]
    .str.replace("%", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce",
)
jointure2["Taux d'occupation des places femmes"] = jointure2["Taux d'occupation des places femmes"]/100
jointure2.loc[10, "Places femmes"] = "2463" #à la main car il y a un pb d'espace
jointure2["Places femmes"] = pd.to_numeric(jointure2["Places femmes"], errors="coerce")
jointure2["Capacité opérationnelle"] = pd.to_numeric(jointure2["Capacité opérationnelle"], errors="coerce")
jointure2["Hommes détenus"] = pd.to_numeric(jointure2["Hommes détenus"], errors="coerce")
```

```python
# calcul du taux d'occupation des places hommes
jointure2["Places hommes"] = jointure2["Capacité opérationnelle"] - jointure2["Places femmes"]
jointure2["Taux d'occupation des places hommes"] = round(
    jointure2["Hommes détenus"] / jointure2["Places hommes"],
    3)

print(jointure2.dtypes)
jointure2
#jointure2.style.format({"Taux d'occupation des places hommes": "{:.0%}", "Taux d'occupation des places femmes": "{:.0%}"})
```

## 2.1 Première visualisation

#### Graphique 1 : Part des femmes parmi les écroués et les détenus

```python
# transformer les valeurs qui sont du texte en nombres 

for col in ["Part de femmes_ecroues", "Part de femmes_detenus"]:
    jointure[col] = (
        jointure[col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.replace("%", "", regex=False)
        .str.strip()
    )
    jointure[col] = pd.to_numeric(jointure[col], errors="coerce")

print(jointure[["Part de femmes_ecroues", "Part de femmes_detenus"]].dtypes)


# créer le graphique

import matplotlib.pyplot as plt

jointure = jointure.sort_values("Au 1er janvier")

#supprimer les lignes avec valeurs manquantes
ecroues = jointure.dropna(subset=["Part de femmes_ecroues"])
detenus = jointure.dropna(subset=["Part de femmes_detenus"])

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(ecroues["Au 1er janvier"], ecroues["Part de femmes_ecroues"], label="Femmes parmi les écroués")
ax.plot(detenus["Au 1er janvier"], detenus["Part de femmes_detenus"], label="Femmes parmi les détenus")

ax.set_title("Part des femmes parmi les écroués et les détenus")
ax.set_xlabel("Année (1er janvier)")
ax.set_ylabel("Part des femmes (%)")
ax.legend()
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

#### Graphique 2

```python
j = jointure3.copy()

j["ratio_ecroues"] = j["Hommes_ecroues"] / j["Femmes_ecroues"]
j["ratio_detenus"] = j["Hommes_detenus"] / j["Femmes_detenus"]

print(j["Catégorie d'âge"].tolist())   # repérer la ligne de total
print(j[["Catégorie d'âge", "ratio_ecroues", "ratio_detenus"]])

ages = j[~j["Catégorie d'âge"].str.contains("ensemble|total", case=False, na=False)]

fig, ax = plt.subplots(figsize=(10, 5))
ages.set_index("Catégorie d'âge")[["ratio_ecroues", "ratio_detenus"]].plot(kind="bar", ax=ax)

ax.set_title("Rapport des taux hommes/femmes par catégorie d'âge")
ax.set_xlabel("Catégorie d'âge")
ax.set_ylabel("Taux hommes / taux femmes")
ax.legend(["Écroués", "Détenus"])
ax.grid(axis="y", alpha=0.3)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Attention, les ratios peuvent être instables dans les tranches d’âge où les taux des femmes sont très faibles (les plus jeunes et les plus âgées)
