Definitions :

- Milieu fermé / ouvert :  une personne relève du milieu fermé lorsqu'elle fait l'objet d'une mesure avec écrou. -  Détenus / non détenus : les détenus sont les personnes écrouées faisant l'objet d'une mesure privative de liberté à l'intérieur d'un établissement pénitentiaire. Les non détenus sont à l'extérieur d'un établissement pénitentiaire.
- L'écrou est une procédure administrative d'identification d'une personne détenue, matérialisée par l'attribution d'un numéro d'écrou au sein de l'établissement pénitentiaire, distinct du titre juridique justifiant l'incarcération. Parmi elleux ya détenus simples, détenus placés à l'extérieur = on leur offre l'hébergement, et condamnés sous bracelet électronique non mobile = on leur offre pas l'hébergement

```python
# jointure tableaux 7, 8 et 12

# tableaux 8 et 7
jointure2 = pd.merge(
    df8_filtre,
    df7_filtre.drop(columns= ["Capacité norme circulaire", "Ecroués détenus"]), 
        # variable inutile (ratio calculé avec capacité opérationnelle) et variable redondante
    on="Direction Interrégionale",
    how="outer",
    validate="one_to_one",
)

# on rajoute tableau 12
jointure2 = pd.merge(
    jointure2,
    df12_filtre.drop(columns=["Etablissement", "Femmes détenues sur places femmes"]), # variable inutile et variable redondante
    on="Direction Interrégionale",
    how="outer",
    validate="one_to_one",
)
```

```python
# Conversion des variables en valeurs numériques
jointure2["Taux d'occupation des places femmes"] = pd.to_numeric(
    jointure2["Taux d'occupation des places femmes"]
    .str.replace("%", "", regex=False)
    .str.replace(",", ".", regex=False),
    errors="coerce",
)
jointure2["Taux d'occupation des places femmes"] = jointure2["Taux d'occupation des places femmes"]/100
jointure2.loc[10, "Places femmes"] = "2463" #à la main car il y a un pb d'espace
jointure2["Places femmes"] = pd.to_numeric(jointure2["Places femmes"], errors="coerce")
jointure2["Capacité opérationnelle"] = pd.to_numeric(jointure2["Capacité opérationnelle"], errors="coerce")
jointure2["Hommes détenus"] = pd.to_numeric(jointure2["Hommes détenus"], errors="coerce")
```

```python
# calcul du taux d'occupation des places hommes
jointure2["Places hommes"] = jointure2["Capacité opérationnelle"] - jointure2["Places femmes"]
jointure2["Taux d'occupation des places hommes"] = round(
    jointure2["Hommes détenus"] / jointure2["Places hommes"],
    3)

print(jointure2.dtypes)
jointure2
#jointure2.style.format({"Taux d'occupation des places hommes": "{:.0%}", "Taux d'occupation des places femmes": "{:.0%}"})
```

```python
import re

def commune(nom):
    nom = re.sub(r"\(.*?\)", "", str(nom)).strip()
    mots = nom.split()
    if mots and mots[0].isupper() and len(mots[0]) <= 4:
        mots = mots[1:]
    return " ".join(mots)

df_carte["commune"] = df_carte["Etablissement"].apply(commune)
print(df_carte[["Etablissement", "commune"]].to_string())
```

```python
import time
from pathlib import Path

import pandas as pd
import requests

URL = "https://data.geopf.fr/geocodage/search"
cache = Path("data/coordonnees_etablissements.csv")

corrections = {
    "BORDEAUX GRADIGNAN": "Gradignan",
    "POITIERS VIVONNE": "Vivonne",
    "LILLE LOOS SEQUEDIN": "Loos",
    "JOUX LA VILLE": "JOUX-LA-VILLE",
    "ORLEANS SARAN": "SARAN",
    "ST ETIENNE LA TALAUDIERE": "LA TALAUDIÈRE",
    "RHONE": "MEYZIEU",
    "LYON CORBAS": "CORBAS",
    "FLEURY MEROGIS": "Fleury-Mérogis",
    "SUD FRANCILIEN": "Réau",
    "CORBEIL ESSONNES": "CORBEIL-ESSONNES",
    "CAEN IFS": "CAEN",
    "MULHOUSE LUTTERBACH": "LUTTERBACH",
    "CHALONS EN CHAMPAGNE": "Châlons-en-Champagne",
    "NANCY MAXEVILLE": "NANCY",
    "TOULOUSE SEYSSES": "SEYSSES",
    "VILLENEUVE LES MAGUELONE": "Villeneuve-lès-Maguelone",
    "BAIE MAHAULT": "Baie-Mahault",
    "FAAA NUUTANIA": "Faaa",
    "MAJICAVO": "KOUNGOU"
}

#  communes, puis corrections
df_carte["commune"] = df_carte["Etablissement"].apply(commune)
df_carte["commune"] = df_carte["commune"].replace(corrections)

absents = [k for k in corrections if k not in df12["Etablissement"].apply(commune).values]
print("Clés de corrections qui ne correspondent à rien :", absents)

# supprimer l'ancien cache (à retirer une fois les coordonnées validées)
cache.unlink(missing_ok=True)


def geocoder(commune):
    r = requests.get(URL, params={"q": commune, "type": "municipality", "limit": 1}, timeout=10)
    r.raise_for_status()
    feats = r.json().get("features", [])
    if not feats:
        return None, None, None, None
    lon, lat = feats[0]["geometry"]["coordinates"]
    p = feats[0]["properties"]
    return lat, lon, p.get("label"), p.get("score")


if cache.exists():
    coords = pd.read_csv(cache)
else:
    lignes = []
    for c in df_carte["commune"].dropna().unique():
        lat, lon, label, score = geocoder(c)
        lignes.append({"commune": c, "lat": lat, "lon": lon, "label": label, "score": score})
        time.sleep(0.1)
    coords = pd.DataFrame(lignes)
    cache.parent.mkdir(parents=True, exist_ok=True)
    coords.to_csv(cache, index=False)

print(coords.sort_values("score").head(10))

# fusion sans doublon de colonnes
df_carte = df_carte.drop(columns=["lat", "lon", "label", "score"], errors="ignore")
df_carte = df_carte.merge(coords, on="commune", how="left")
print(df_carte["lat"].isna().sum(), "établissements sans coordonnées")
```
